```python
# Import necessary libraries
import pandas as pd
import yfinance as yf
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
import datetime as dt
```

```python
# Define the function to download stock data
def download_stock_data(tickers, start_date, end_date):
    stock_data = {}
    for ticker in tickers:
        stock_data[ticker] = yf.download(ticker, start=start_date, end=end_date)
    return stock_data
```

```python
# Define the function to calculate daily returns
def calculate_daily_returns(stock_data):
    daily_returns = {}
    for ticker, data in stock_data.items():
        data['Daily Return'] = data['Adj Close'].pct_change() * 100
        daily_returns[ticker] = data['Daily Return']
    return pd.DataFrame(daily_returns)
```

```python
# Define the function to calculate volatility (standard deviation)
def calculate_volatility(daily_returns):
    volatility = daily_returns.std() * np.sqrt(252)  # Annualized volatility
    return volatility
```

```python
# Define the function to plot stock prices
def plot_stock_prices(stock_data):
    plt.figure(figsize=(10, 6))
    for ticker, data in stock_data.items():
        plt.plot(data.index, data['Adj Close'], label=ticker)
    plt.title('Stock Prices Over Time')
    plt.xlabel('Date')
    plt.ylabel('Adjusted Close Price')
    plt.legend()
    plt.show()
```

```python
# Define the function to plot returns distribution
def plot_returns_distribution(daily_returns):
    plt.figure(figsize=(10, 6))
    sns.histplot(daily_returns, kde=True, bins=50)
    plt.title('Distribution of Daily Returns')
    plt.xlabel('Daily Return (%)')
    plt.show()
```

```python
# Define the function to calculate the Sharpe Ratio
def calculate_sharpe_ratio(daily_returns, risk_free_rate=0.02):
    mean_returns = daily_returns.mean() * 252  # Annualize mean return
    volatility = calculate_volatility(daily_returns)
    sharpe_ratios = (mean_returns - risk_free_rate) / volatility
    return sharpe_ratios
```

```python
# Main function to execute the analysis
def stock_market_analysis(tickers, start_date, end_date):
    # Step 1: Download stock data
    stock_data = download_stock_data(tickers, start_date, end_date)
    
    # Step 2: Plot stock prices
    plot_stock_prices(stock_data)
    
    # Step 3: Calculate daily returns
    daily_returns = calculate_daily_returns(stock_data)
    
    # Step 4: Plot returns distribution
    plot_returns_distribution(daily_returns)
    
    # Step 5: Calculate and display volatility and Sharpe ratios
    volatility = calculate_volatility(daily_returns)
    
    sharpe_ratios = calculate_sharpe_ratio(daily_returns)
    
    # Create a summary DataFrame
    summary_df = pd.DataFrame({
        'Volatility (%)': volatility,
        'Sharpe Ratio': sharpe_ratios
    })
    
    print("\nStock Analysis Summary:")
    print(summary_df)
```

```python

# Input parameters
tickers = ["AAPL", "MSFT", "GOOG", "AMZN", "TSLA", "NVDA", "META", "ORCL", "ADBE", "CRM"]  # Example stock tickers
# Time define
years = 1

end_date = dt.datetime.now()
start_date = end_date - dt.timedelta(days = 365 * years)

# Execute the analysis
stock_market_analysis(tickers, start_date, end_date)

# Define a dictionary of stocks categorized by sector
'''
stocks = {
    "Technology": ["AAPL", "MSFT", "GOOG", "AMZN", "TSLA", "NVDA", "META", "ORCL", "ADBE", "CRM"],
    "Healthcare": ["JNJ", "PFE", "MRNA", "UNH", "ABT", "LLY", "GILD", "BMY", "CVS", "ISRG"],
    "Financials": ["JPM", "BAC", "WFC", "GS", "C", "MS", "AXP", "BRK.B", "BLK", "V"],
    "Consumer Discretionary": ["HD", "NKE", "SBUX", "MCD", "LOW", "TGT", "TSLA", "AMZN", "F", "GM"],
    "Energy": ["XOM", "CVX", "BP", "COP", "PSX", "SLB", "OXY", "HAL", "MPC", "KMI"],
    "Utilities": ["NEE", "DUK", "SO", "D", "EXC", "AEP", "PPL", "XEL", "ES", "AWK"],
    "Consumer Staples": ["PG", "KO", "PEP", "WMT", "COST", "MO", "PM", "CL", "KMB", "MDLZ"],
    "Industrials": ["HON", "BA", "LMT", "GE", "MMM", "CAT", "DE", "RTX", "GD", "UPS"],
    "Materials": ["LIN", "APD", "SHW", "ECL", "DD", "FCX", "NEM", "VMC", "MLM", "ALB"],
    "Real Estate": ["AMT", "PLD", "CCI", "SPG", "EQIX", "PSA", "WELL", "DLR", "O", "VTR"],
    "Communication Services": ["T", "VZ", "TMUS", "DIS", "CHTR", "NFLX", "GOOG", "META", "CMCSA", "TWTR"]
}
'''
```
